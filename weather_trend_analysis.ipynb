```python

# Load temperature dataset

df = pd.read_csv("temperature.csv")

# Display first 5 rows
df.head()
```

```python

# Check dataset information

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum().head(10))
```

```python

# Convert datetime into date format

df['datetime'] = pd.to_datetime(df['datetime'])

# Check data types
print(df['datetime'].dtype)
```

```python

print("Total Missing Values:")
print(df.isnull().sum().sum())
```

```python

# Fill missing temperature values

df.iloc[:, 1:] = df.iloc[:, 1:].ffill()

# Check missing values again
print(df.isnull().sum().sum())
```

```python

# Select New York temperature data

city = 'New York'

city_data = df[['datetime', city]]

print(city_data.head())
```

```python

df.iloc[:, 1:] = df.iloc[:, 1:].ffill().bfill()

print("Missing values:", df.isnull().sum().sum())
```

```python

city = 'New York'

city_data = df[['datetime', city]].copy()

# Convert Kelvin to Celsius
city_data[city] = city_data[city] - 273.15

print(city_data.head())
```

```python

plt.figure(figsize=(14, 5))

plt.plot(city_data['datetime'], city_data['New York'])

plt.title('New York Temperature Trend (2012–2017)')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')

plt.grid(True)
plt.show()
```

```python

city_data['Year'] = city_data['datetime'].dt.year
city_data['Month'] = city_data['datetime'].dt.month

print(city_data.head())
```

```python

monthly_avg = city_data.groupby('Month')['New York'].mean()

print(monthly_avg)
```

```python

plt.figure(figsize=(10, 5))

plt.plot(monthly_avg.index, monthly_avg.values, marker='o')

plt.title('Average Monthly Temperature in New York')
plt.xlabel('Month')
plt.ylabel('Temperature (°C)')

plt.xticks(range(1, 13))
plt.grid(True)
plt.show()
```

```python

print("Hottest Month:", monthly_avg.idxmax())
print("Highest Temperature:", round(monthly_avg.max(), 2), "°C")

print("Coldest Month:", monthly_avg.idxmin())
print("Lowest Temperature:", round(monthly_avg.min(), 2), "°C")
```

```python

yearly_avg = city_data.groupby('Year')['New York'].mean()

print(yearly_avg)
```

```python

plt.figure(figsize=(12, 5))

plt.plot(yearly_avg.index, yearly_avg.values, marker='o')

plt.title('Yearly Average Temperature in New York')
plt.xlabel('Year')
plt.ylabel('Temperature (°C)')

plt.grid(True)
plt.show()
```

```python

print("Hottest Year:", yearly_avg.idxmax())
print("Highest Average:", round(yearly_avg.max(), 2), "°C")

print("Coldest Year:", yearly_avg.idxmin())
print("Lowest Average:", round(yearly_avg.min(), 2), "°C")
```

```python

cities = ['New York', 'London', 'Paris', 'Tokyo', 'San Francisco']

# Check which cities are available
print([c for c in cities if c in df.columns])
```

```python

cities = ['New York', 'Chicago', 'Miami',
          'Los Angeles', 'San Francisco']

city_avg = df[cities].mean() - 273.15

print(city_avg)
```

```python

plt.figure(figsize=(10, 5))

city_avg.sort_values().plot(kind='bar')

plt.title('Average Temperature Comparison Across Cities')
plt.xlabel('Cities')
plt.ylabel('Average Temperature (°C)')

plt.xticks(rotation=45)
plt.show()
```

```python

humidity_df = pd.read_csv("humidity.csv")

print("Dataset Shape:", humidity_df.shape)

humidity_df.head()
```

```python

# Convert datetime
humidity_df['datetime'] = pd.to_datetime(humidity_df['datetime'])

# Fill missing values using interpolation
humidity_df.iloc[:, 1:] = humidity_df.iloc[:, 1:].interpolate(
    limit_direction='both'
)

print("Missing values:",
      humidity_df.isnull().sum().sum())
```

```python

humidity_data = humidity_df[['datetime', 'New York']].copy()

humidity_data['Year'] = humidity_data['datetime'].dt.year
humidity_data['Month'] = humidity_data['datetime'].dt.month

print(humidity_data.head())
```

```python

monthly_humidity = humidity_data.groupby('Month')['New York'].mean()

print(monthly_humidity)
```

```python

plt.figure(figsize=(10, 5))

plt.plot(monthly_humidity.index,
         monthly_humidity.values, marker='o')

plt.title('Average Monthly Humidity in New York')
plt.xlabel('Month')
plt.ylabel('Humidity (%)')

plt.xticks(range(1, 13))
plt.grid(True)
plt.show()
```

```python

print("Highest Humidity Month:", monthly_humidity.idxmax())
print("Highest Humidity:", round(monthly_humidity.max(), 2), "%")

print("Lowest Humidity Month:", monthly_humidity.idxmin())
print("Lowest Humidity:", round(monthly_humidity.min(), 2), "%")
```

```python
wind_df = pd.read_csv("wind_speed.csv")

print("Dataset Shape:", wind_df.shape)
wind_df.head()
```

```python
wind_df['datetime'] = pd.to_datetime(wind_df['datetime'])

wind_df.iloc[:, 1:] = wind_df.iloc[:, 1:].interpolate(
    limit_direction='both'
)

print("Missing values:", wind_df.isnull().sum().sum())
```

```python
wind_data = wind_df[['datetime', 'New York']].copy()

wind_data['Year'] = wind_data['datetime'].dt.year
wind_data['Month'] = wind_data['datetime'].dt.month

print(wind_data.head())
```

```python
monthly_wind = wind_data.groupby('Month')['New York'].mean()

print(monthly_wind)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(monthly_wind.index, monthly_wind.values, marker='o')

plt.title('Average Monthly Wind Speed in New York')
plt.xlabel('Month')
plt.ylabel('Wind Speed')
plt.xticks(range(1, 13))
plt.grid(True)

plt.show()
```

```python
print("Windiest Month:", monthly_wind.idxmax())
print("Highest Average Wind Speed:",
      round(monthly_wind.max(), 2))

print("Least Windy Month:", monthly_wind.idxmin())
print("Lowest Average Wind Speed:",
      round(monthly_wind.min(), 2))
```
